MLOps Assignment 2: Track Machine Learning Experiments Using MLflow

This notebook explores the Wine dataset before training. The actual model training and MLflow logging is done in `src/train.py` (run it from the terminal so MLflow runs are tracked correctly).

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/wine.csv')
df.columns = [c.strip() for c in df.columns]
df.head()
````

````python
print(df.shape)
print(df.isnull().sum())
print(df['class'].value_counts())
````

````python
df.describe()
````

````python
plt.figure(figsize=(6,4))
sns.countplot(x='class', data=df)
plt.title('Class Distribution')
plt.show()
````

````python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
````
